```python
import re

def normalize_text(text):
    # Convert to lowercase
    text = text.lower()

    # Remove punctuation (except apostrophes)
    text = re.sub(r'[^\w\s\']', '', text)

    # Remove extra whitespace
    text = ' '.join(text.split())

    return text

# Define a function to add the normalized_text column
def add_normalized_text(example):
    example['normalized_text'] = normalize_text(example['text'])
    return example

# Apply the function to the dataset
dataset = dataset.map(add_normalized_text)

# Print the first few examples to verify
print(dataset[2:5])
```

```python
def extract_all_chars(batch):
    all_text = " ".join(batch["normalized_text"])
    vocab = list(set(all_text))
    return {"vocab": [vocab], "all_text": [all_text]}


vocabs = dataset.map(
    extract_all_chars,
    batched=True,
    batch_size=-1,
    keep_in_memory=True,
    remove_columns=dataset.column_names,
)

dataset_vocab = set(vocabs["vocab"][0])
tokenizer_vocab = {k for k, _ in tokenizer.get_vocab().items()}
```

```python
import os
import torch
from speechbrain.pretrained import EncoderClassifier

spk_model_name = "speechbrain/spkrec-xvect-voxceleb"

device = "cuda" if torch.cuda.is_available() else "cpu"
speaker_model = EncoderClassifier.from_hparams(
    source=spk_model_name,
    run_opts={"device": device},
    savedir=os.path.join("/tmp", spk_model_name),
)


def create_speaker_embedding(waveform):
    with torch.no_grad():
        speaker_embeddings = speaker_model.encode_batch(torch.tensor(waveform))
        speaker_embeddings = torch.nn.functional.normalize(speaker_embeddings, dim=2)
        speaker_embeddings = speaker_embeddings.squeeze().cpu().numpy()
    return speaker_embeddings
```

```python
def prepare_dataset(example):
    audio = example["audio"]

    example = processor(
        text=example["normalized_text"],
        audio_target=audio["array"],
        sampling_rate=audio["sampling_rate"],
        return_attention_mask=False,
    )

    # strip off the batch dimension
    example["labels"] = example["labels"][0]

    # use SpeechBrain to obtain x-vector
    example["speaker_embeddings"] = create_speaker_embedding(audio["array"])

    return example
```

```python
from dataclasses import dataclass
from typing import Any, Dict, List, Union


@dataclass
class TTSDataCollatorWithPadding:
    processor: Any

    def __call__(
        self, features: List[Dict[str, Union[List[int], torch.Tensor]]]
    ) -> Dict[str, torch.Tensor]:
        input_ids = [{"input_ids": feature["input_ids"]} for feature in features]
        label_features = [{"input_values": feature["labels"]} for feature in features]
        speaker_features = [feature["speaker_embeddings"] for feature in features]

        # collate the inputs and targets into a batch
        batch = processor.pad(
            input_ids=input_ids, labels=label_features, return_tensors="pt"
        )

        # replace padding with -100 to ignore loss correctly
        batch["labels"] = batch["labels"].masked_fill(
            batch.decoder_attention_mask.unsqueeze(-1).ne(1), -100
        )

        # not used during fine-tuning
        del batch["decoder_attention_mask"]

        # round down target lengths to multiple of reduction factor
        if model.config.reduction_factor > 1:
            target_lengths = torch.tensor(
                [len(feature["input_values"]) for feature in label_features]
            )
            target_lengths = target_lengths.new(
                [
                    length - length % model.config.reduction_factor
                    for length in target_lengths
                ]
            )
            max_length = max(target_lengths)
            batch["labels"] = batch["labels"][:, :max_length]

        # also add in the speaker embeddings
        batch["speaker_embeddings"] = torch.tensor(speaker_features)

        return batch
```

```python
import os
import librosa
import torch
import numpy as np
from datasets import load_dataset
from transformers import (
    Wav2Vec2Processor, Wav2Vec2ForCTC,
    SpeechT5Processor, SpeechT5ForTextToSpeech, SpeechT5HifiGan
)
from jiwer import wer, cer
from scipy.io.wavfile import write

# -------------------------------
# 1️⃣ Load LJSpeech Dataset (Ground Truth)
# -------------------------------
ljspeech = load_dataset("lj_speech", split="train")  # Use "train" for full dataset

# Create ground truth dictionary {filename: text}
ground_truths = {f"LJ{item['id']}.wav": item["text"].strip().lower() for item in ljspeech}

# Directory to save generated audio
generated_audio_dir = "generated_audio"
os.makedirs(generated_audio_dir, exist_ok=True)

# -------------------------------
# 2️⃣ Load Pretrained TTS Model (SpeechT5) & Speaker Embeddings
# -------------------------------
tts_model = SpeechT5ForTextToSpeech.from_pretrained("microsoft/speecht5_tts")
tts_processor = SpeechT5Processor.from_pretrained("microsoft/speecht5_tts")
vocoder = SpeechT5HifiGan.from_pretrained("microsoft/speecht5_hifigan")

# Load speaker embeddings (initialize with zeros for consistency)
speaker_embeddings = torch.zeros(1, 512)

# -------------------------------
# 3️⃣ Generate & Save TTS Audio
# -------------------------------
def generate_audio(text, output_path):
    """Generate speech from text using SpeechT5 and save as a WAV file."""
    if not isinstance(text, str) or not text.strip():
        raise ValueError(f"Invalid text input for TTS: {repr(text)}")

    print(f"🎙️ Generating speech for: {repr(text)}")

    # Convert text to tensor for SpeechT5
    inputs = tts_processor(text=[text], return_tensors="pt")  # Ensure text is in a list

    with torch.no_grad():
        speech = tts_model.generate(**inputs, speaker_embeddings=speaker_embeddings)
        speech_waveform = vocoder(speech).cpu().numpy().squeeze()

    # Save as WAV file
    write(output_path, rate=16000, data=(speech_waveform * 32767).astype(np.int16))

# Generate first 5 samples
print("\n🎙️ Generating SpeechT5 Audio...")
for i, (filename, text) in enumerate(ground_truths.items()):
    if i >= 5: break  # Generate for only 5 samples
    output_path = os.path.join(generated_audio_dir, filename)
    generate_audio(text, output_path)
    print(f"✅ Generated: {filename}")

from transformers import Wav2Vec2Processor

# -------------------------------
# 4️⃣ Load ASR Model (Wav2Vec2) for Transcription
# -------------------------------
# Load Wav2Vec2 model with a language model (KenLM)
asr_processor = Wav2Vec2Processor.from_pretrained("facebook/wav2vec2-large-960h")
asr_model = Wav2Vec2ForCTC.from_pretrained("facebook/wav2vec2-large-960h")


# -------------------------------
# 5️⃣ Transcribe Audio (Ground Truth & Generated)
# -------------------------------
def transcribe_audio(file_path):
    """Transcribe audio using Wav2Vec2 with beam search decoding."""
    speech, sr = librosa.load(file_path, sr=16000)
    input_values = asr_processor(speech, return_tensors="pt", sampling_rate=16000).input_values

    with torch.no_grad():
        logits = asr_model(input_values).logits

    # Decode and get the first element of the list
    predicted_text_list = asr_processor.batch_decode(torch.argmax(logits, dim=-1))
    predicted_text = predicted_text_list[0] if predicted_text_list else ""

    return predicted_text.strip().lower()

# -------------------------------
# 6️⃣ Evaluate WER & CER
# -------------------------------
total_wer, total_cer = 0, 0
num_samples = 0

print("\n🔍 Evaluating ASR Transcriptions...")
for filename in os.listdir(generated_audio_dir):
    if filename in ground_truths:
        gen_audio_path = os.path.join(generated_audio_dir, filename)

        # Transcribe both ground truth & generated audio
        ground_truth_text = ground_truths[filename]
        gen_transcription = transcribe_audio(gen_audio_path)

        # Compute WER & CER
        sample_wer = wer(ground_truth_text, gen_transcription)
        sample_cer = cer(ground_truth_text, gen_transcription)

        # Print results
        print(f"\n📢 File: {filename}")
        print(f"🔹 Ground Truth: {ground_truth_text}")
        print(f"🔹 Transcribed: {gen_transcription}")
        print(f"✅ WER: {sample_wer:.4f}, CER: {sample_cer:.4f}")

        total_wer += sample_wer
        total_cer += sample_cer
        num_samples += 1

# -------------------------------
# 7️⃣ Print Overall Scores
# -------------------------------
if num_samples > 0:
    avg_wer = total_wer / num_samples
    avg_cer = total_cer / num_samples
    print(f"\n🚀 Overall WER: {avg_wer:.4f}, Overall CER: {avg_cer:.4f}")
else:
    print("No valid test samples found.")
```

```python
import os
import librosa
import torch
import numpy as np
from datasets import load_dataset
from transformers import (
    Wav2Vec2ProcessorWithLM, Wav2Vec2ForCTC,
    SpeechT5Processor, SpeechT5ForTextToSpeech, SpeechT5HifiGan
)
from jiwer import wer, cer
from scipy.io.wavfile import write

# -------------------------------
# 1️⃣ Load LJSpeech Dataset (Ground Truth)
# -------------------------------
ljspeech = load_dataset("lj_speech", split="train")
ground_truths = {f"LJ{item['id']}.wav": item["text"].strip().lower() for item in ljspeech}

# Directory to save generated audio
output_dirs = {
    "pretrained": "generated_audio_pretrained",
    "finetuned": "generated_audio_finetuned",
    "enhanced": "generated_audio_enhanced",
}
for d in output_dirs.values():
    os.makedirs(d, exist_ok=True)

# -------------------------------
# 2️⃣ Load SpeechT5 Models and their corresponding Processors
# -------------------------------
tts_models = {
    "pretrained": SpeechT5ForTextToSpeech.from_pretrained("microsoft/speecht5_tts"),
    "finetuned": SpeechT5ForTextToSpeech.from_pretrained("farazashraf/speecht5_ljspeech_finetuned"),
    "enhanced": SpeechT5ForTextToSpeech.from_pretrained("farazashraf/speecht5_finetuned_enhanced"),
}

# Use the corresponding processor for each model
tts_processors = {
    "pretrained": SpeechT5Processor.from_pretrained("microsoft/speecht5_tts"),
    "finetuned": SpeechT5Processor.from_pretrained("farazashraf/speecht5_ljspeech_finetuned"),
    "enhanced": SpeechT5Processor.from_pretrained("farazashraf/speecht5_finetuned_enhanced"),
}

vocoder = SpeechT5HifiGan.from_pretrained("microsoft/speecht5_hifigan")
speaker_embeddings = torch.zeros(1, 512)  # Zero embeddings for uniform testing

# -------------------------------
# 3️⃣ Generate & Save SpeechT5 Audio using the matching processor
# -------------------------------
def generate_audio(model, processor, text, output_path):
    inputs = processor(text=[text], return_tensors="pt")
    with torch.no_grad():
        speech = model.generate(**inputs, speaker_embeddings=speaker_embeddings)
        speech_waveform = vocoder(speech).cpu().numpy().squeeze()
    write(output_path, rate=16000, data=(speech_waveform * 32767).astype(np.int16))

print("\n🎙️ Generating SpeechT5 Audio...")
for model_name, model in tts_models.items():
    processor = tts_processors[model_name]
    for i, (filename, text) in enumerate(ground_truths.items()):
        if i >= 5:
            break  # Generate only 5 samples
        output_path = os.path.join(output_dirs[model_name], filename)
        generate_audio(model, processor, text, output_path)
        print(f"✅ {model_name.capitalize()} Model Generated: {filename}")

# -------------------------------
# 4️⃣ Load ASR Model (Wav2Vec2) for Transcription
# -------------------------------
from transformers import Wav2Vec2Processor

asr_processor = Wav2Vec2Processor.from_pretrained("facebook/wav2vec2-large-960h")
asr_model = Wav2Vec2ForCTC.from_pretrained("facebook/wav2vec2-large-960h")

# -------------------------------
# 5️⃣ Transcribe Audio (Ground Truth & Generated)
# -------------------------------
def transcribe_audio(file_path):
    """Transcribe audio using Wav2Vec2 with beam search decoding."""
    speech, sr = librosa.load(file_path, sr=16000)
    input_values = asr_processor(speech, return_tensors="pt", sampling_rate=16000).input_values

    with torch.no_grad():
        logits = asr_model(input_values).logits

    # Decode and get the first element of the list
    predicted_text_list = asr_processor.batch_decode(torch.argmax(logits, dim=-1))
    predicted_text = predicted_text_list[0] if predicted_text_list else ""

    return predicted_text.strip().lower()

# -------------------------------
# 5️⃣ Evaluate WER & CER
# -------------------------------
results = {}
print("\n🔍 Evaluating ASR Transcriptions...")
for model_name, model_dir in output_dirs.items():
    total_wer, total_cer, num_samples = 0, 0, 0

    for filename in os.listdir(model_dir):
        if filename in ground_truths:
            file_path = os.path.join(model_dir, filename)
            gt_text = ground_truths[filename]
            transcribed_text = transcribe_audio(file_path)
            sample_wer, sample_cer = wer(gt_text, transcribed_text), cer(gt_text, transcribed_text)

            print(f"\n📢 {model_name.capitalize()} Model - File: {filename}")
            print(f"🔹 Ground Truth: {gt_text}")
            print(f"🔹 Transcribed: {transcribed_text}")
            print(f"✅ WER: {sample_wer:.4f}, CER: {sample_cer:.4f}")

            total_wer += sample_wer
            total_cer += sample_cer
            num_samples += 1

    avg_wer, avg_cer = (total_wer / num_samples, total_cer / num_samples) if num_samples > 0 else (0, 0)
    results[model_name] = {"WER": avg_wer, "CER": avg_cer}

# -------------------------------
# 6️⃣ Print Final Comparison
# -------------------------------
print("\n🚀 SpeechT5 Model Comparison Results:")
for model_name, scores in results.items():
    print(f"{model_name.capitalize()} Model - WER: {scores['WER']:.4f}, CER: {scores['CER']:.4f}")
```

```python
import os
import librosa
import torch
import time
import numpy as np
from datasets import load_dataset
from transformers import (
    SpeechT5Processor, SpeechT5ForTextToSpeech, SpeechT5HifiGan
)
from scipy.io.wavfile import write
from fastdtw import fastdtw
from scipy.spatial.distance import euclidean

# Set device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# -------------------------------
# 1️⃣ Load LJSpeech Dataset (Ground Truth Audio & Text)
# -------------------------------
ljspeech = load_dataset("lj_speech", split="train")

# Extract GT audio directly from the dataset
ground_truths = {}
gt_audio_samples = {}

for item in ljspeech.select(range(5)):  # Limit to first 5 samples
    filename = f"LJ{item['id']}.wav"
    ground_truths[filename] = item["text"].strip().lower()
    gt_audio_samples[filename] = item["audio"]["array"]

# Directory to save generated audio
output_dirs = {
    "pretrained": "generated_audio_pretrained",
    "finetuned": "generated_audio_finetuned",
    "enhanced": "generated_audio_enhanced",
}
for d in output_dirs.values():
    os.makedirs(d, exist_ok=True)

# -------------------------------
# 2️⃣ Load SpeechT5 Models (CPU)
# -------------------------------
tts_models = {
    "pretrained": SpeechT5ForTextToSpeech.from_pretrained("microsoft/speecht5_tts"),
    "finetuned": SpeechT5ForTextToSpeech.from_pretrained("farazashraf/speecht5_ljspeech_finetuned"),
    "enhanced": SpeechT5ForTextToSpeech.from_pretrained("farazashraf/speecht5_finetuned_enhanced"),
}

# Move TTS models to device
for key, model in tts_models.items():
    tts_models[key] = model.to(device)

tts_processors = {
    "pretrained": SpeechT5Processor.from_pretrained("microsoft/speecht5_tts"),
    "finetuned": SpeechT5Processor.from_pretrained("farazashraf/speecht5_ljspeech_finetuned"),
    "enhanced": SpeechT5Processor.from_pretrained("farazashraf/speecht5_finetuned_enhanced"),
}
vocoder = SpeechT5HifiGan.from_pretrained("microsoft/speecht5_hifigan").to(device)
speaker_embeddings = torch.zeros(1, 512).to(device)  # Zero embeddings for uniform testing

# -------------------------------
# 3️⃣ Generate & Save SpeechT5 Audio + Measure Inference Time
# -------------------------------
def generate_audio(model, processor, text, output_path):
    inputs = processor(text=[text], return_tensors="pt")
    inputs = {k: v.to(device) for k, v in inputs.items()}
    start_time = time.time()

    with torch.no_grad():
        speech = model.generate(**inputs, speaker_embeddings=speaker_embeddings)
        speech_waveform = vocoder(speech).cpu().numpy().squeeze()

    inference_time = (time.time() - start_time) * 1000  # Convert to ms
    write(output_path, rate=16000, data=(speech_waveform * 32767).astype(np.int16))
    return inference_time

print("\n🎙️ Generating SpeechT5 Audio & Measuring Performance...")
metrics = {model: {"RTF": [], "MCD": [], "Inference": []} for model in tts_models}

for model_name, model in tts_models.items():
    processor = tts_processors[model_name]
    for filename, text in ground_truths.items():
        output_path = os.path.join(output_dirs[model_name], filename)

        inference_time = generate_audio(model, processor, text, output_path)
        generated_speech, sr = librosa.load(output_path, sr=16000)

        # Compute RTF (Real-Time Factor)
        real_time_factor = inference_time / (len(generated_speech) / sr * 1000)

        # Store metrics
        metrics[model_name]["RTF"].append(real_time_factor)
        metrics[model_name]["Inference"].append(inference_time)

        print(f"✅ {model_name.capitalize()} Model - {filename}")
        print(f"⏳ Inference Time: {inference_time:.2f} ms, ⚡ RTF: {real_time_factor:.4f}")

# -------------------------------
# 4️⃣ Compute Mel-Cepstral Distortion (MCD)
# -------------------------------
def compute_mcd(gt_audio, gen_audio):
    gt_mfcc = librosa.feature.mfcc(y=gt_audio, sr=16000, n_mfcc=13)
    gen_mfcc = librosa.feature.mfcc(y=gen_audio, sr=16000, n_mfcc=13)
    distance, _ = fastdtw(gt_mfcc.T, gen_mfcc.T, dist=euclidean)
    return distance / len(gt_mfcc.T)  # Normalize

print("\n🔍 Computing MCD...")
for model_name, model_dir in output_dirs.items():
    for filename in os.listdir(model_dir):
        if filename in ground_truths:
            gt_audio = gt_audio_samples[filename]  # Directly get GT audio from dataset
            gen_audio, _ = librosa.load(os.path.join(model_dir, filename), sr=16000)

            mcd_score = compute_mcd(gt_audio, gen_audio)
            metrics[model_name]["MCD"].append(mcd_score)
            print(f"📊 {model_name.capitalize()} Model - {filename} - MCD: {mcd_score:.4f}")

# -------------------------------
# 5️⃣ Compute Final Averages & Print Results
# -------------------------------
print("\n🚀 SpeechT5 Model Performance Comparison:")

for model_name, scores in metrics.items():
    avg_rtf = np.mean(scores["RTF"])
    avg_mcd = np.mean(scores["MCD"])
    avg_inference = np.mean(scores["Inference"])

    print(f"\n📢 {model_name.capitalize()} Model:")
    print(f"🔹 Real-Time Factor (RTF): {avg_rtf:.4f} ↓")
    print(f"🔹 Mel-Cepstral Distortion (MCD): {avg_mcd:.4f} ↓")
    print(f"🔹 Inference Time: {avg_inference:.2f} ms ↓")
```
